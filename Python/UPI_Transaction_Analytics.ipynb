```python
#Extract Date Features

df['date'] = df['timestamp'].dt.date
df['year'] = df['timestamp'].dt.year
df['month'] = df['timestamp'].dt.month
df["month_name"] = df["timestamp"].dt.month_name()
df['day'] = df['timestamp'].dt.day
```

```python
df[['timestamp',"year","month","month_name","day"]].head()
```

```python
#Create Time Slot

def time_slot(hour):
    if 0 <= hour < 6:
        return "Night"
    elif 6 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 18:
        return "Afternoon"
    else:
        return "Evening"

df["time_slot"] = df["hour_of_day"].apply(time_slot)
```

```python
df["time_slot"].value_counts()
```

```python
#Amount Category

def amount_category(amount):
    if amount < 500:
        return "Low"
    elif amount < 5000:
        return "Medium"
    elif amount < 20000:
        return "High"
    else:
        return "Very High"

df["amount_category"] = df["amount_inr"].apply(amount_category)
```

```python
df["amount_category"].value_counts()
```

```python
#Weekend Label
df["week_type"] = df["is_weekend"].map({
    True: "Weekend",
    False: "Weekday"
})
```
